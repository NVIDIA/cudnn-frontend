# Causal Conv1D NWH Forward Operation

This notebook shows how to compute a causal depthwise 1D convolution forward operation using cuDNN with NWH (batch, seqLen, dim) layout — also known as "NWH".

$$y = \text{Activation}(\text{CausalConv1D}(x, w) + b)$$

where Activation is Identity or SiLU.

The NWH layout stores the channel (dim) dimension last, which is the natural memory order for transformer-style models where the hidden dimension is the innermost axis. The convolution uses left-only (causal) padding of size `kernel_size - 1`, and each channel is convolved independently (depthwise).

**Layout comparison:**

| | NHW (notebook 60) | NWH (this notebook) |
|---|---|---|
| x, y | `(batch, dim, seq_len)` | `(batch, seq_len, dim)` |
| weight | `(dim, kernel_size)` | `(kernel_size, dim)` |
| bias | `(dim,)` | `(dim,)` |

## Overview

We will perform a causal depthwise conv1d forward pass with NWH layout:

- batch size: 2
- dim (channels): 768
- sequence length: 4096
- kernel size: 4
- activation: SiLU
- data type: BFloat16

We compare the cuDNN result against a PyTorch reference implementation.

````python
import torch
import torch.nn.functional as F
import cudnn

print("cuDNN backend version:", cudnn.backend_version())
````

## Reference Implementation

A simple PyTorch reference for causal depthwise conv1d + SiLU with NWH layout.

````python
def causal_conv1d_nwh_ref(x, weight, bias, activation="silu"):
    """Reference: causal depthwise conv1d for NWH layout using PyTorch ops.

    Args:
        x:      (batch, seq_len, dim)
        weight: (kernel_size, dim)
        bias:   (dim,)
        activation: 'silu' or 'identity'

    Returns:
        y: (batch, seq_len, dim)
    """
    batch, seq_len, dim = x.shape
    kernel_size = weight.shape[0]

    # Transpose to NHW for PyTorch conv1d
    x_nhw = x.transpose(1, 2)  # (batch, dim, seq_len)

    # NWH weight (kernel_size, dim) → NHW weight (dim, 1, kernel_size) for groups=dim
    w_nhw = weight.T.unsqueeze(1)  # (dim, 1, kernel_size)

    # Causal padding: (kernel_size - 1) on the left, 0 on the right
    x_padded = F.pad(x_nhw, (kernel_size - 1, 0))
    y = F.conv1d(x_padded, w_nhw, bias=bias, groups=dim)

    if activation == "silu":
        y = F.silu(y)

    return y.transpose(1, 2)  # Back to NWH
````

## Setup Tensors

````python
batch = 2
dim = 768
seq_len = 4096
kernel_size = 4
dtype = torch.bfloat16

has_cuda = torch.cuda.is_available()

torch.manual_seed(42)

x = torch.randn(batch, seq_len, dim, dtype=dtype)
weight = torch.randn(kernel_size, dim, dtype=dtype)
bias = torch.randn(dim, dtype=dtype)

print(f"CUDA available: {has_cuda}")
print(f"x:      {x.shape}, dtype={x.dtype}")
print(f"weight: {weight.shape}, dtype={weight.dtype}")
print(f"bias:   {bias.shape}, dtype={bias.dtype}")
````

## Compute Reference Output

````python
y_ref = causal_conv1d_nwh_ref(x, weight, bias, activation="silu")
print(f"y_ref:  {y_ref.shape}, dtype={y_ref.dtype}")
````

## Compute cuDNN Output

````python
if has_cuda:
    x_gpu = x.cuda()
    weight_gpu = weight.cuda()
    bias_gpu = bias.cuda()
    y_cudnn = cudnn.ops.causal_conv1d_nwh(x_gpu, weight_gpu, bias=bias_gpu, activation="silu")
    print(f"y_cudnn: {y_cudnn.shape}, dtype={y_cudnn.dtype}")
else:
    print("Skipping cuDNN forward (no CUDA device). Run on a GPU machine to test cuDNN.")
````

## Verify Correctness

````python
if has_cuda:
    y_cudnn_cpu = y_cudnn.cpu()
    max_abs_diff = (y_cudnn_cpu.float() - y_ref.float()).abs().max().item()
    max_rel_diff = ((y_cudnn_cpu.float() - y_ref.float()).abs() / (y_ref.float().abs() + 1e-6)).max().item()

    print(f"Max absolute difference: {max_abs_diff:.6e}")
    print(f"Max relative difference: {max_rel_diff:.6e}")

    atol = 1e-2
    rtol = 1e-2

    assert torch.allclose(y_cudnn_cpu.float(), y_ref.float(), atol=atol, rtol=rtol), \
        f"FAILED: max_abs={max_abs_diff}, max_rel={max_rel_diff}"

    print("PASSED: cuDNN causal_conv1d NWH forward matches reference.")
else:
    print("Skipping verification (no CUDA device).")
````

## Test with Different Data Types and Kernel Sizes

````python
tolerances = {
    torch.float32:  (5e-6, 5e-6),
    torch.float16:  (5e-3, 5e-3),
    torch.bfloat16: (1e-2, 1e-2),
}

test_configs = [
    {"dtype": torch.float32,  "kernel_size": 4,  "activation": "silu"},
    {"dtype": torch.float16,  "kernel_size": 4,  "activation": "silu"},
    {"dtype": torch.bfloat16, "kernel_size": 4,  "activation": "silu"},
    {"dtype": torch.bfloat16, "kernel_size": 3,  "activation": "silu"},
    {"dtype": torch.float16,  "kernel_size": 7,  "activation": "silu"},
    {"dtype": torch.bfloat16, "kernel_size": 4,  "activation": "identity"},
]

batch, dim, seq_len = 2, 256, 1024

for cfg in test_configs:
    dt = cfg["dtype"]
    ks = cfg["kernel_size"]
    act = cfg["activation"]
    atol, rtol = tolerances[dt]

    x = torch.randn(batch, seq_len, dim, dtype=dt)
    w = torch.randn(ks, dim, dtype=dt)
    b = torch.randn(dim, dtype=dt)

    y_ref = causal_conv1d_nwh_ref(x, w, b, activation=act)

    if has_cuda:
        y_cudnn = cudnn.ops.causal_conv1d_nwh(x.cuda(), w.cuda(), bias=b.cuda(), activation=act)
        max_abs = (y_cudnn.cpu().float() - y_ref.float()).abs().max().item()
        passed = torch.allclose(y_cudnn.cpu().float(), y_ref.float(), atol=atol, rtol=rtol)
        status = "PASS" if passed else "FAIL"
        print(f"[{status}] dtype={str(dt):15s} K={ks:3d} act={act:10s} max_abs_diff={max_abs:.4e}")
        assert passed, f"Test failed for {cfg}"
    else:
        print(f"[REF ] dtype={str(dt):15s} K={ks:3d} act={act:10s} y_ref shape={y_ref.shape}")

print("\nAll tests passed!" if has_cuda else "\nAll reference tests completed (CPU only).")
````

## torch.compile Support

`cudnn.ops.causal_conv1d_nwh` is registered as a PyTorch custom operator, so it works seamlessly with `torch.compile`.

````python
if has_cuda:
    B, D, L, K = 2, 256, 1024, 4
    x_c = torch.randn(B, L, D, dtype=torch.bfloat16, device="cuda")
    w_c = torch.randn(K, D, dtype=torch.bfloat16, device="cuda")
    b_c = torch.randn(D, dtype=torch.bfloat16, device="cuda")

    @torch.compile
    def compiled_causal_conv1d_nwh(x, w, b):
        return cudnn.ops.causal_conv1d_nwh(x, w, bias=b, activation="silu")

    y_eager = cudnn.ops.causal_conv1d_nwh(x_c, w_c, bias=b_c, activation="silu")
    y_compiled = compiled_causal_conv1d_nwh(x_c, w_c, b_c)

    max_abs = (y_compiled.float() - y_eager.float()).abs().max().item()
    print(f"torch.compile matches eager: max_abs_diff={max_abs:.4e}")
    assert max_abs == 0.0, f"torch.compile output differs from eager: {max_abs}"
    print(f"y_compiled: {y_compiled.shape}, dtype={y_compiled.dtype}")
    print("PASSED: torch.compile produces identical output to eager mode.")
else:
    print("Skipping torch.compile test (no CUDA device).")
````
